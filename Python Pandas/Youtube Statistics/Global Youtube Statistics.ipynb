```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df=pd.read_csv('Global Youtube Statistics.csv',encoding='Latin 1')
```

```python
df.set_index('rank',inplace=True)
```

```python
df
```

```python
df.info()
```

```python
df.columns
```

```python
df.dropna(axis=0,how='any',subset=['created_year','created_month','created_date','Country of origin','subscribers'],inplace=True)
df
```

```python
df.info()
```

```python
df.fillna({'Category':'Unknown','channel_type':'Unknown','video_views_for_the_last_30_days':0,'subscribers_for_last_30_days':0},inplace=True)
df
```

```python
group_by_category_df=df.groupby(by='category').agg({'subscribers':sum,'video views':sum}).sort_values(by=['subscribers','video views'],ascending=False)
group_by_category_df
```

```python
fig=plt.figure(figsize=(10,10))
ax=fig.add_subplot()
group_by_category_df.plot(kind='bar',stacked=True)
```
